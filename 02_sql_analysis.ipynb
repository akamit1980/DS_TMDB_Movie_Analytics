```python
#List all movies in the Action genre released after 2015, showing title and release date.

query = """
    SELECT title, release_date 
    FROM movies 
    JOIN movie_genres ON movies.movie_id = movie_genres.movie_id 
    JOIN genres ON movie_genres.genre_id = genres.genre_id 
    WHERE genres.genre_name = %s 
      AND YEAR(movies.release_date) > 2015
"""

cursor.execute(query, ("Action",))
for data in cursor:
  print(data)
```

```python
#Calculate the average budget and average revenue for each genre, sorted by average revenue descending.

query = """
SELECT genre_name, AVG(budget) AS avg_budget, AVG(revenue) AS avg_revenue 
FROM movies 
JOIN movie_genres ON movies.movie_id = movie_genres.movie_id 
JOIN genres ON movie_genres.genre_id = genres.genre_id 
GROUP BY genre_name 
ORDER BY avg_revenue DESC;
"""

cursor.execute(query)

for data in cursor:
  print(data)
```

```python
# List the top 10 actors who have appeared in the most movies in the dataset.(cast)

query = """
SELECT actor_name, count(movie_id) FROM cast group by actor_name
LIMIT 10;
"""

cursor.execute(query)
for data in cursor:
    print(data)
```

```python
#Find all directors who have directed more than 3 movies, along with their movie count.

query = """
        SELECT person_name, job, COUNT(movie_id) 
        FROM crew 
        WHERE job IN (SELECT job FROM crew WHERE job = 'director') 
        GROUP BY job, person_name 
        HAVING COUNT(*) > 3;
        """

cursor.execute(query)
for data in cursor:
    print(data)
```

```python
# List the top 10 most frequently used keywords across all movies.

query = """
        SELECT keyword_name, COUNT(movie_id) 
        FROM movie_keywords
        GROUP BY keyword_id, keyword_name
        LIMIT 10;
        """

cursor.execute(query)
for data in cursor:
    print(data)
```

```python
# Find movies where the budget was above the overall average budget but the revenue was below the overall average revenue.

query = """
SELECT title, budget, revenue 
FROM movies 
WHERE budget > (SELECT AVG(budget) FROM movies)
  AND revenue < (SELECT AVG(revenue) FROM movies);  
"""

cursor.execute(query)
for data in cursor:
    print(data)
```

```python
# List all actors who have appeared in a movie directed by a specific director (e.g., Christopher Nolan).

query = """
SELECT DISTINCT actor_name 
FROM `cast` as c
JOIN crew as cr on c.movie_id = cr.movie_id
WHERE cr.job= 'Director' and cr.person_name = "Christopher Nolan";
"""

cursor.execute(query)
for data in cursor:
    print(data)
```

```python
# Find the genre with the highest average rating, considering only genres with at least 20 movies.
# Movie, movie_genres and genres

query = """
SELECT 
    m.movie_id, 
    m.title, 
    vote_average
FROM movies m
JOIN movie_genres mg ON m.movie_id = mg.movie_id
WHERE mg.genre_id IN (
    SELECT genre_id 
    FROM movie_genres 
    GROUP BY genre_id 
    HAVING COUNT(*) >= 20
)
GROUP BY m.movie_id, m.title
ORDER BY vote_average DESC
LIMIT 20;
"""
    
cursor.execute(query)
for data in cursor:
    print(data)
```

```python
# movie and movie_keywords
# group by movie Id and title
# List the top 10 movies with the highest number of keywords tagged to them.

query = """
            SELECT 
                m.title, 
                COUNT(mk.keyword_id) AS keyword_count
            FROM 
                movies m
            INNER JOIN 
                movie_keywords mk ON m.movie_id = mk.movie_id
            GROUP BY 
                m.movie_id, m.title
            ORDER BY 
                keyword_count DESC
            LIMIT 10;
            """
            
cursor.execute(query)
for data in cursor:
    print(data) 
```
