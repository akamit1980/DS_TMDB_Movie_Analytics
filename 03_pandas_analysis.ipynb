```python
# 2. List all movies with missing or zero budget/revenue values, and decide how these should be treated for the rest of the analysis.

import pandas as pd
import numpy as np
df = pd.read_csv(r'C:\AIML Course\Data\Project1\movies_cleaned.csv')
df['release_date'] = pd.to_datetime(df['release_date'])
df = df.astype({'budget': int, 'revenue': int})

df['budget'] = df['budget'].replace(0, np.nan)
df['revenue'] = df['revenue'].replace(0, np.nan)
zeros_in_column = df[df['budget'] == 0]
zeros_in_column = df[df['revenue'] == 0]
print(zeros_in_column)
df['budget'] = df['budget'].fillna(df['budget'].median())
df['revenue'] = df['revenue'].fillna(df['revenue'].median())

df.drop_duplicates()
df.to_csv(r'C:\AIML Course\Data\Project1\movies_cleaned.csv', index=False)
df.isnull().sum()
df.info()
zero_counts = (df == 0).sum()
print(zero_counts)
```

```python
# 3. List the profit (revenue − budget) and return on investment for every movie that has valid budget and revenue figures.

valid_movies = df[
    df['budget'].notna() & (df['budget'] > 0) & 
    df['revenue'].notna() & (df['revenue'] > 0)
].copy()
valid_movies['profit'] = valid_movies['revenue'] - valid_movies['budget']
valid_movies['roi'] = valid_movies['profit'] / valid_movies['budget']
columns_to_show = ['title', 'budget', 'revenue', 'profit', 'roi']
result = valid_movies[columns_to_show]
print(result)
```

```python
#4.  List the top 15 movies by return on investment, considering only movies with a budget of at least $1 million.

df = df[df["budget"] >= 1000000].copy()

df["roi"] = ((df["revenue"] - df["budget"]) / df["budget"]) * 100

top_15_movies = df.sort_values(by="roi", ascending=False).head(15)
print(top_15_movies[["title", "budget", "revenue", "roi"]])
```

```python
# 5. List the number of movies released per year, and identify which year had the highest movie output in the dataset.

import pandas as pd
df = pd.read_csv(r"C:\AIML Course\Data\Project1\movies_cleaned.csv")

df['Release Date'] = pd.to_datetime(df['release_date'])
movies_per_year = df['Release Date'].dt.year.value_counts().sort_index()

print("Number of movies released per year:")
print(movies_per_year)

highest_year = movies_per_year.idxmax()
highest_count = movies_per_year.max()

print(f"\nYear with the highest movie output: {highest_year} ({highest_count} movies)")
```

```python
# 6. List all movies whose runtime is a statistical outlier compared to the rest of the dataset.

import pandas as pd

df = pd.read_csv(r"C:\AIML Course\Data\Project1\movies_cleaned.csv")

Q1 = df["runtime"].quantile(0.25)
Q3 = df["runtime"].quantile(0.75)
IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR
outliers_df = df[(df["runtime"] < lower_limit) | (df["runtime"] > upper_limit)]

print(f"Lower Limit: {lower_limit} minutes")
print(f"Upper Limit: {upper_limit} minutes")
print(f"Total Outlier Movies Found: {len(outliers_df)}\n")

print(outliers_df[["title", "runtime"]].sort_values(by="runtime"))
```

```python
# 7. List each movie's primary genre (the first genre listed) and compare average popularity across primary genres.

#1. merge movies_geners and genres
#2. sort new dataframe by using movie_id and genre Id and then group by movie id
#3. again we need to merge second dataframe with movies 
#4. group by genre and then display the popularity

import pandas as pd

df = pd.read_csv(r'C:\AIML Course\Data\Project1\movie_genres_cleaned.csv')
df1 = pd.read_csv(r'C:\AIML Course\Data\Project1\genres_cleaned.csv')
df2 = pd.read_csv(r'C:\AIML Course\Data\Project1\movies_cleaned.csv')

merged_df = pd.merge(df, df1, on='genre_id', how='inner')

merged_df_sorted = merged_df.sort_values(by=['movie_id', 'genre_id']).groupby('movie_id').first().reset_index()
new_df = pd.merge(merged_df, df2, on='movie_id', how='inner' )
genre_popularity = new_df.groupby('genre_name')['popularity'].mean().reset_index()

print(genre_popularity)
```

```python
# 8. List the top 10 most prolific actors by number of movies, along with their average movie rating across those movies.
#merge cast with movies
#grupby actor and diplay aggregate for vote_average head 10

import pandas as pd
df = pd.read_csv(r'C:\AIML Course\Data\Project1\cast_cleaned.csv')
df1 = pd.read_csv(r'C:\AIML Course\Data\Project1\movies_cleaned.csv')

new_df = pd.merge(df, df1, on='movie_id', how='inner')

actor_stats = new_df.groupby('actor_name')['vote_average'].agg(['mean', 'median', 'count'])

actor_stats.columns = ['Avg Rating', 'Median Rating', 'Movie Count']

actor_stats = actor_stats.sort_values(by='Avg Rating', ascending=False)

actor_stats.head(10)


```

```python
# 9. List any movie titles that appear more than once in the dataset, along with their release years, to check for duplicates or remakes.
df = pd.read_csv(r"C:\AIML Course\Data\Project1\movies_cleaned.csv")


df['release_date'] = pd.to_datetime(df['release_date'], errors='coerce')


df['release_year'] = df['release_date'].dt.year


duplicates_all = df[df.duplicated(subset=["title"], keep=False)]


duplicate_titles = duplicates_all.sort_values(
    by=["title", "release_year"]
)[["title", "release_year"]]

print(duplicate_titles)
```

```python
# 10. List movies whose popularity score and vote count don't follow the general pattern between the two — i.e., unusually high popularity with a low vote count, or vice versa.

import pandas as pd

df = pd.read_csv(r"C:\AIML Course\Data\Project1\movies_cleaned.csv")

df['popularity_rank'] = df['popularity'].rank(ascending=False)

df['vote_count_rank'] = df['vote_count'].rank(ascending=False)

df['rank_gap'] = df['popularity_rank'] - df['vote_count_rank']

df_sorted = df.sort_values(by='rank_gap', ascending=False)

df_sorted[['title', 'popularity_rank', 'vote_count_rank', 'rank_gap']]
```
